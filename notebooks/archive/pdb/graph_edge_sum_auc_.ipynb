```python
from torch_geometric.data import OnDiskDataset


pdb_dset = OnDiskDataset("../../data/pdb/triad/graphs/pdb_triad_af3")
```

```python
from torch_geometric.utils import to_dense_adj, get_laplacian
import torch
import numpy as np

y_true = []
y_score = []

for graph in pdb_dset:

    # sum edges from peptide to everything else
    segids = np.array(graph.segids)

    # last peptide res
    last_peptide_res = np.argwhere(segids == "A")[:, 0][-1] + 1

    # all outgoing edges
    outgoing_edges = graph.edge_index[0] < last_peptide_res

    # outgoing contact probs
    outgoing_probs = graph.edge_attr[outgoing_edges, 1]

    y_true.append(graph.y[0])

    y_score.append(outgoing_probs.sum())
```

```python
from sklearn.metrics import roc_auc_score

roc_auc_score(y_true, y_score)
```

```python
pdb_dset.close()
```
