```python
import polars as pl
import numpy as np
import pprint

np.set_printoptions(precision=3)
pdb_af3 = pl.read_parquet("../../data/pdb/triad/staged/pdb_triad.af3_rmsd.parquet")
```

```python
def fmt(ls):
    return np.array2string(
        np.array(ls),
        separator=", ",
        formatter={"float_kind": lambda x: f"{x:.3f}"},
    ).replace("\n", "")
```

```python
worst_cdr_rmsd_I = list(
    pdb_af3.filter(pl.col("mhc_class") == "I")
    .sort(by="cdr_rmsd_af3_4", descending=True)[0]
    .iter_rows(named=True)
)[0]
print(worst_cdr_rmsd_I["pdb"])
print(worst_cdr_rmsd_I["job_name"])
print("Predicted TCR geom:")
print(
    "origin="
    + fmt(worst_cdr_rmsd_I["pred_tcr_origin_4"])
    + ", vectors="
    + fmt(worst_cdr_rmsd_I["pred_tcr_axes_4"])
)
print("Predicted MHC origin:")
print(
    "origin="
    + fmt(worst_cdr_rmsd_I["pred_mhc_origin_4"])
    + ", vectors="
    + fmt(worst_cdr_rmsd_I["pred_mhc_axes_4"])
)

print("True TCR geom:")
print(
    "origin="
    + fmt(worst_cdr_rmsd_I["true_tcr_origin"])
    + ", vectors="
    + fmt(worst_cdr_rmsd_I["true_tcr_axes"])
)
print("True MHC geom:")
print(
    "origin="
    + fmt(worst_cdr_rmsd_I["true_mhc_origin"])
    + ", vectors="
    + fmt(worst_cdr_rmsd_I["true_mhc_axes"])
)
```

```python
best_cdr_rmsd_I = list(
    pdb_af3.filter(pl.col("mhc_class") == "I")
    .sort(by="cdr_rmsd_af3_4", descending=False)[0]
    .iter_rows(named=True)
)[0]
print(best_cdr_rmsd_I["pdb"])
print(best_cdr_rmsd_I["job_name"])
print("Predicted TCR geom:")
print(
    "origin="
    + fmt(best_cdr_rmsd_I["pred_tcr_origin_4"])
    + ", vectors="
    + fmt(best_cdr_rmsd_I["pred_tcr_axes_4"])
)
print("Predicted MHC origin:")
print(
    "origin="
    + fmt(best_cdr_rmsd_I["pred_mhc_origin_4"])
    + ", vectors="
    + fmt(best_cdr_rmsd_I["pred_mhc_axes_4"])
)

print("True TCR geom:")
print(
    "origin="
    + fmt(best_cdr_rmsd_I["true_tcr_origin"])
    + ", vectors="
    + fmt(best_cdr_rmsd_I["true_tcr_axes"])
)
print("True MHC geom:")
print(
    "origin="
    + fmt(best_cdr_rmsd_I["true_mhc_origin"])
    + ", vectors="
    + fmt(best_cdr_rmsd_I["true_mhc_axes"])
)
```
